### Perceptron
感知机 - 神经网络（深度学习）的起源的算法

#### 感知机
感知机接受多个输入信号，输出一个信号（信号只有 1/0 两种）</br>
图2.1是例子:
* x1、x2是输入信号，y是输出信号，w1、w2是权重。
* 图中的○称为“神经元”或者“节点”
* 权重越大，对应该权重的信号的重要性就 越高。
> 输入信号被送往神经元时，会被分别乘以固定的权重 （w1x1、w2x2）。神经元会计算传送过来的信号的总和，只有当这个总和**超过了某个界限值**时，才会输出1。这也称为“神经元被激活” 。这里将这个界限值称为阈值，用符号θ表示。

$y= \begin{cases} 0 & (w_1x_1+w_2x_2\leqslant\theta) \\ 1 & (w_1x_1+w_2x_2>\theta) & \end{cases}\quad(2.1)$

#### 简单逻辑电路
* 与门 AND
* 与非门 NAND -- Not AND
* 或门 OR

3个门电路只有参数的值（权重和阈值）不同</br>
相同构造的感知机，只用改变权重w1, w2, θ就可以变成不同的门
> 相同构造的感知机，只需通过适当地调整参数的值，就可以像“变色龙演员”表演不 同的角色一样，变身为与门、与非门、或门。

#### 感知机的实现
下面代码是与门，只需要把参数取倒数就变成了与非门

```python
def AND(x1, x2):
    w1, w2, theta = 0.5, 0.5, 0.7
    tmp = x1 * w1 + x2 * w2
    if(tmp > theta):
        return 1
    elif(tmp <= theta):
        return 0

print(AND(0, 0))
print(AND(0, 1))
print(AND(1, 0))
print(AND(1, 1))
```

#### 导入权重和偏置
先把式（2.1）的θ换成−b <br>
$y= \begin{cases} 0 & (b+w_1x_1+w_2x_2\leqslant0) \\ 1 & (b+w_1x_1+w_2x_2>0) & \end{cases}\quad(2.2)$ <br>
b称为偏置，w1和w2称为权重。

```python
import numpy as np
x = np.array([0, 1])
w = np.array([0.5, 0.5])
b = -0.7
print(w * x)
print(np.sum(w * x))
print(np.sum(w * x) + b) # 浮点小数造成的运算误差
```

#### 使用偏置实现
偏置和权重w1 、w2 的作用是不一样的
* w1 和 w2 是控制输入信号的重要性的参数
* 而偏置 b 是调整神经元被激活的容易程度（输出信号为1的程度）的参数

```python
def AND2(x1, x2):
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.7
    tmp = np.sum(x * w) + b
    if(tmp > 0):
        return 1
    else:
        return 0
    
def NAND(x1, x2):
    x = np.array([x1, x2])
    w = np.array([-0.5, -0.5])
    b = 0.7
    tmp = np.sum(x * w) + b
    if(tmp > 0):
        return 1
    else:
        return 0
    
def OR(x1, x2):
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.2
    tmp = np.sum(x * w) + b
    if(tmp > 0):
        return 1
    else:
        return 0
    
print(OR(0, 0))
print(OR(0, 1))
print(OR(1, 0))
print(OR(1, 1))
```

#### 类

```python
class test:
    def __init__(self):
        pass # 这里是构造函数，只在生成类的实例调用一次
    def fun1(self):
        print("fun1") # 方法一
    def fun2(self):
        print("fun2") # 方法二


class Man:
    def __init__(self, name):
        self.name = name
        print("Initialized!")

    def hello(self):
        print("hello " + self.name + "!")

    def goodbye(self):
        print("Good-bye " + self.name + "!")

m = Man("Tom")
m.hello()
m.goodbye()
```

### Numpy

#### 数组

```python
import numpy as np
x = np.array([1.0, 2.0, 3.0])
print(x)
type(x)
```

#### 计算

```python
x = np.array([1, 2, 3])
y = np.array([2, 4, 6])
print(x + y) # 元素不同会报错
print(x - y)
print(x * y) # element-wise product 对应元素的乘法
print(x / y)

print(x / 2) # 广播 数组与标量做计算
```

#### N维数组

```python
A = np.array([[1, 2], [3, 4]])
print(A)
print(A.shape) # 数组形状
print(A.dtype) # 数组数据类型

print("--------------------")
B = np.array([[3, 0], [0, 6]])
print(A + B)
print(A * B) # 对应元素相乘
print(A * 10)# 广播
```

#### 广播
形状不同的数组之间也可以进行运算

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([10, 20])
A * B       # 维度对应相乘
```

#### 访问元素

```python
X = np.array([[51, 55], [14, 19], [0, 4]])
print(X)
print(X[0])     # 第0行
print(X[0][1])  # 具体位置数据

for row in X:   # 循环访问
    print(row)

print("------------------------")
X = X.flatten() # 将X转换成一维数组
print(X)
print(X[np.array([0, 2, 4])]) # 三个位置的数据

print("------------------------")
print(X > 15)   # 比大小
print(X[X > 15])
```

### Matplotlib

#### 绘图

```python
import numpy as np
import matplotlib.pyplot as plt

# 生成数据
x = np.arange(0, 6, 0.1) # 以 0.1 为单位，生成0到6的数据
y = np.sin(x)

# 绘图
plt.plot(x, y)
plt.show()
```

#### pyplot功能

```python
import numpy as np
import matplotlib.pyplot as plt

# 生成数据
x = np.arange(0, 6, 0.1) # 以0.1为单位，生成0-6的数据
y1 = np.sin(x)
y2 = np.cos(x)

# 绘制图形
plt.plot(x, y1, label = "sin")
plt.plot(x, y2, linestyle = "--", label = 'cos') # 用虚线回执
plt.xlabel("X") # X轴坐标
plt.ylabel("Y") # y轴坐标
plt.title('sin & cos') # 标题
plt.legend()    # label显示
plt.show()
```
